# 04 — Export
Exportar apenas o adapter é a opção recomendada na L4. Merge exige RAM de sistema suficiente e carrega o modelo-base completo em CPU.

```python
from pathlib import Path
import os, subprocess, sys
PROJECT_ROOT = Path(os.environ.get('FABLE_PROJECT_ROOT', '/content/logos-3')).resolve()
os.chdir(PROJECT_ROOT)
ADAPTER = Path('outputs/adapters/stage1')
OUTPUT = Path('/content/drive/MyDrive/fable-qwen-distillation/adapters/stage1-export')
MERGE = False
command = [sys.executable, 'scripts/export_model.py', '--adapter', str(ADAPTER), '--output', str(OUTPUT)]
if MERGE:
    command.append('--merge')
subprocess.run(command, check=True)
```
