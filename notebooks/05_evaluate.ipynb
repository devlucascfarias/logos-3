# 05 — Evaluate
Calcule métricas estáticas para predições geradas com os mesmos seeds/parâmetros. O harness aceita tarefas próprias com `repository_path` e uma lista de comandos já gerada/validada.

```python
from pathlib import Path
import os, subprocess, sys
PROJECT_ROOT = Path(os.environ.get('FABLE_PROJECT_ROOT', '/content/logos-3')).resolve()
os.chdir(PROJECT_ROOT)
PREDICTIONS = Path('outputs/evaluations/stage1_predictions.jsonl')
if PREDICTIONS.exists():
    subprocess.run([sys.executable, 'scripts/evaluate_static.py', '--config', 'configs/eval.yaml', '--predictions', str(PREDICTIONS), '--output', 'outputs/evaluations/stage1_static.json'], check=True)
else:
    print('Gere predições base/stage1/stage2 com os mesmos parâmetros e salve em', PREDICTIONS)
```

```python
TASKS = Path('data/processed/agent_tasks.jsonl')
if TASKS.exists():
    subprocess.run([sys.executable, 'scripts/evaluate_agent.py', '--tasks', str(TASKS), '--output', 'outputs/evaluations/agent_runs.jsonl'], check=True)
else:
    print('Adicione tarefas executáveis próprias em', TASKS)
```
