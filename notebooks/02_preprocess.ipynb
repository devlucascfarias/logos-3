# 02 — Preprocess
Normaliza as fontes, cria modalidades de treino, deduplica e divide por grupos conectados. Comece com `SOURCE_LIMIT=750` para a Fase 0.

```python
from pathlib import Path
import os, subprocess, sys
PROJECT_ROOT = Path(os.environ.get('FABLE_PROJECT_ROOT', '/content/logos-3')).resolve()
os.chdir(PROJECT_ROOT)
SOURCE_LIMIT = 750
build = [sys.executable, 'scripts/build_examples.py', '--config', 'configs/data.yaml']
if SOURCE_LIMIT:
    build += ['--limit', str(SOURCE_LIMIT), '--max-output-examples', str(SOURCE_LIMIT)]
subprocess.run(build, check=True)
subprocess.run([sys.executable, 'scripts/deduplicate.py', '--input', 'data/interim/examples.jsonl', '--output', 'data/interim/examples.dedup.jsonl'], check=True)
subprocess.run([sys.executable, 'scripts/split_by_group.py', '--input', 'data/interim/examples.dedup.jsonl', '--output-dir', 'data/processed'], check=True)
subprocess.run([sys.executable, '-m', 'pytest', '-q'], check=True)
```
